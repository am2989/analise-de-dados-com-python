```python

import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("student-mat.csv", sep=";")

plt.figure(figsize=(6, 6))
plt.hist(df['absences'], bins=20, edgecolor='black')
plt.title('Distribuição da variável Faltas')
plt.xlabel('Faltas')
plt.ylabel('Nota Final')
plt.grid(axis='y', alpha=0.75)
plt.show()

#    A maioria dos estudantes tem pouquíssimas faltas — o primeiro intervalo tem quase 200 ocorrências.
#À medida que o número de faltas aumenta, a frequência cai de forma acentuada.
#A distribuição é assimétrica à direita (skewed right): muitos com poucas faltas, poucos com muitas faltas.
#Interpretação estatística Isto indica que faltas elevadas são raras. A variável “faltas” tem uma cauda longa, típica de comportamentos onde a maioria cumpre e poucos se desviam muito.
```

```python
#usar este grafico
plt.hist(df['absences'], bins=20)
plt.title("Histograma das Faltas")

#**Resumo direto:** o gráfico mostra que a maioria das pessoas tem **poucas faltas** e que **faltas muito altas são raras**. É uma distribuição **assimétrica à direita** (muitos valores baixos, poucos valores altos).


## O que este **histograma** revela Concentração perto do zero:** a maior parte dos alunos tem entre **0 e 5 faltas**.
#**Cauda longa à direita:** existem alguns casos com muitas faltas (20, 30, 40, até ~70), mas são **poucos**.
# **Distribuição não é uniforme:** há um padrão claro de que **quanto maior o número de faltas, menor a frequência**.
# **Possível presença de outliers:** valores muito altos podem indicar **casos excepcionais**.


## 🧠 Interpretação estatística

#A variável **não é normal**; é **assimétrica positiva**.
# A média provavelmente é **maior que a mediana**, porque a cauda puxa a média para cima.
# A maior parte dos dados está concentrada no início do eixo X → **baixa variabilidade para a maioria**, mas **alta variabilidade nos casos extremos**.

---

## 🔍 O que podes concluir sobre o comportamento dos alunos

#A maioria falta pouco → comportamento geral **regular**.
#Há um pequeno grupo com faltas muito elevadas → pode indicar **problemas específicos**, como desmotivação, dificuldades pessoais ou risco de retenção.
```

```python
#usar este gráfico
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
var = "address"
sns.boxplot(x=var, y='G3', data=df.dropna(subset=['G3', var]), showfliers=False)
plt.title('Boxplot das Notas Finais por Residência')
plt.xlabel('Tipo de Residência')
plt.ylabel('Nota Final (G3)')
plt.show()

#Este boxplot compara a nota final (G3) entre alunos de residência: U → urbana R → rural
#1. Mediana das notas A linha no centro da caixa representa a mediana. Em U, a mediana está perto de 11. Em R, está perto de 10.

#Isto indica que os alunos urbanos tendem a ter notas finais ligeiramente superiores.

#2.Dispersão das notas A altura da caixa mostra a variabilidade das notas. O grupo R parece ter uma dispersão um pouco maior.
#Isso significa que entre os alunos rurais existem diferenças maiores entre notas baixas e altas.
#3. Valores mínimos e máximos

#Os “bigodes” mostram os extremos normais.

#Em U, as notas variam aproximadamente entre 4 e 20.
#Em R, variam entre 0 e 19.
#O grupo rural apresenta casos de nota muito baixa (0), enquanto no urbano o mínimo é mais alto.

#4. Sobreposição das caixas

#As caixas sobrepõem-se bastante. Isso significa que: os dois grupos têm distribuições relativamente semelhantes;a diferença existe, mas não é muito forte.
#Conclusão geral O gráfico sugere que: alunos urbanos (U) têm desempenho final ligeiramente melhor; alunos rurais (R) apresentam maior variabilidade nas notas;
#a residência pode ter alguma influência no G3, mas não é uma diferença muito grande.#
```

```python
#usar este grafico
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.scatterplot(x='absences', y='G3', data=df,
                hue='address',
                style='address',
                alpha=0.7
                )
plt.title('Scatterplot de Faltas vs. Nota Final por Residência')
plt.xlabel('Faltas')
plt.ylabel('Nota Final')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(title='Residência')
plt.show()

#Este scatterplot mostra a relação entre: Faltas (absences) → eixo X Nota final (G3) → eixo Y Tipo de residência (address) azul = urbano (U) laranja = rural (R)

#1. Tendência negativa entre faltas e nota À medida que as faltas aumentam: as notas tendem a diminuir.Isto sugere uma correlação negativa: mais faltas → pior desempenho académico.
#2. Concentração principal dos alunos A maior parte dos pontos está: entre 0 e 15 faltas e notas entre 8 e 16

#Isso indica que a maioria dos alunos: tem relativamente poucas faltas; e notas médias/positivas.
#3. Muitos alunos com poucas faltas têm boas notas No lado esquerdo do gráfico: vemos vários alunos com: poucas faltas, e notas entre 14 e 20.

#Isso reforça a ideia de que assiduidade ajuda no desempenho.

#4. Alunos com muitas faltas tendem a ter notas baixas

#À direita do gráfico: aparecem alunos com 30, 40, 50 e até 75 faltas; quase todos têm notas mais baixas.

#Não aparecem notas muito altas associadas a muitas faltas.

#5. Comparação entre urbano (U) e rural (R) Os dois grupos estão bastante misturados.

#Isso significa que: o padrão entre faltas e nota é parecido nos dois tipos de residência; o tipo de residência não parece alterar muito a relação. 6. Existem exceções

#O gráfico também mostra alguns casos fora da tendência: alunos com poucas faltas mas notas baixas; alguns com faltas moderadas e notas razoáveis.

#Portanto, as faltas influenciam o desempenho, mas não explicam tudo.

#Conclusão geral O gráfico indica que: existe uma relação negativa entre faltas e nota final; alunos com menos faltas tendem a obter melhores resultados;
#o padrão é semelhante para alunos urbanos e rurais;
#a residência tem pouca influência comparada com o impacto das faltas.
```

Interpretação:

Através da análise dos gráficos, foi possível identificar padrões relevantes no que respeita à zona de residência dos alunos, a quantidade de faltas registadas e o seu desempenho académico.

Os dados revelam que a maioria dos alunos se enquadra na categoria "U", enquanto a categoria "R" corresponde a uma porção bem menor do total de estudantes. Isto evidencia uma presença muito significativa de alunos da zona urbana.

Conclusão:
